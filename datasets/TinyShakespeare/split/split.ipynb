```python
import os
```

```python
with open("../tiny_shakespeare.txt", "r") as f:
    data = f.read()  # Read the entire file as a single string
```

```python
# Calculate split indices
total_length = len(data)
train_size = int(0.8 * total_length)
val_size = int(0.1 * total_length)
test_size = total_length - train_size - val_size

# Split the data
train_data = data[:train_size]
val_data = data[train_size:train_size + val_size]
test_data = data[train_size + val_size:]
```

```python
# Save training data
with open("train.txt", "w") as f:
    f.write(train_data)

# Save validation data
with open("val.txt", "w") as f:
    f.write(val_data)

# Save testing data
with open("test.txt", "w") as f:
    f.write(test_data)
```

```python
# Verify file sizes
print(f"Training data size: {len(train_data)}")
print(f"Validation data size: {len(val_data)}")
print(f"Testing data size: {len(test_data)}")

# Optionally, print the first few lines of each file
with open("train.txt", "r") as f:
    print("Training data sample:", f.read(100))  # Print first 100 characters

with open("val.txt", "r") as f:
    print("Validation data sample:", f.read(100))

with open("test.txt", "r") as f:
    print("Testing data sample:", f.read(100))
```
